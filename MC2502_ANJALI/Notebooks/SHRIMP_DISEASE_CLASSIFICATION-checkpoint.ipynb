```python
import os
import random
import copy
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

from PIL import Image

import torch
import torch.nn as nn
from torch.utils.data import Dataset, DataLoader, WeightedRandomSampler

from torchvision import transforms
from torchvision.models import resnet18, ResNet18_Weights

from sklearn.metrics import (
    accuracy_score,
    precision_recall_fscore_support,
    classification_report,
    confusion_matrix,
    ConfusionMatrixDisplay,
    roc_auc_score
)

print("All imports successful!")
```

```python
SEED = 42

random.seed(SEED)
np.random.seed(SEED)
torch.manual_seed(SEED)

device = torch.device("cuda" if torch.cuda.is_available() else "cpu")

print("Device:", device)
```

```python
train_transform = transforms.Compose([
    transforms.RandomResizedCrop(224, scale=(0.75, 1.0)),
    transforms.RandomHorizontalFlip(p=0.5),
    transforms.RandomRotation(15),
    transforms.ColorJitter(
        brightness=0.2,
        contrast=0.2,
        saturation=0.2
    ),
    transforms.ToTensor(),
    transforms.Normalize(
        mean=[0.485, 0.456, 0.406],
        std=[0.229, 0.224, 0.225]
    )
])
```

```python
eval_transform = transforms.Compose([
    transforms.Resize(256),
    transforms.CenterCrop(224),
    transforms.ToTensor(),
    transforms.Normalize(
        mean=[0.485, 0.456, 0.406],
        std=[0.229, 0.224, 0.225]
    )
])
```

```python
class FishShrimpBinaryDataset(Dataset):
    
    def __init__(
        self,
        hf_dataset,
        transform=None,
        species_filter=None
    ):
        
        self.data = hf_dataset
        self.transform = transform
        self.species_filter = species_filter
        
        self.indices = []
        
        for i in range(len(self.data)):
            
            label_name = self.data[i]["label_name"]
            species = get_species(label_name)
            
            if species_filter is None or species == species_filter:
                self.indices.append(i)
    
    def __len__(self):
        return len(self.indices)
    
    def __getitem__(self, idx):
        
        real_idx = self.indices[idx]
        
        item = self.data[real_idx]
        
        # Convert image to RGB
        image = item["image"].convert("RGB")
        
        # Binary label
        label = binary_label(item["label_name"])
        
        # Apply transformation
        if self.transform:
            image = self.transform(image)
        
        return image, torch.tensor(
            label,
            dtype=torch.long
        )
```

```python
stage1_train = FishShrimpBinaryDataset(
    train_hf,
    transform=train_transform
)

stage1_valid = FishShrimpBinaryDataset(
    valid_hf,
    transform=eval_transform
)

stage1_test = FishShrimpBinaryDataset(
    test_hf,
    transform=eval_transform
)

print("Stage 1 Train:", len(stage1_train))
print("Stage 1 Valid:", len(stage1_valid))
print("Stage 1 Test :", len(stage1_test))
```

```python
def get_binary_labels(dataset):
    
    labels = []
    
    for idx in dataset.indices:
        label_name = dataset.data[idx]["label_name"]
        labels.append(binary_label(label_name))
    
    return np.array(labels)


stage1_train_labels = get_binary_labels(stage1_train)

class_counts = np.bincount(
    stage1_train_labels,
    minlength=2
)

print("Class counts:")
print("Healthy  :", class_counts[0])
print("Diseased :", class_counts[1])


class_weights = 1.0 / class_counts

sample_weights = torch.DoubleTensor(
    class_weights[stage1_train_labels]
)

sampler = WeightedRandomSampler(
    weights=sample_weights,
    num_samples=len(sample_weights),
    replacement=True
)

print("Weighted sampler created successfully!")
```

```python
def get_binary_labels(dataset):
    
    labels = []
    
    for idx in dataset.indices:
        label_name = dataset.data[idx]["label_name"]
        labels.append(binary_label(label_name))
    
    return np.array(labels)


stage1_train_labels = get_binary_labels(stage1_train)

class_counts = np.bincount(
    stage1_train_labels,
    minlength=2
)

print("Class counts:")
print("Healthy  :", class_counts[0])
print("Diseased :", class_counts[1])


class_weights = 1.0 / class_counts

sample_weights = torch.DoubleTensor(
    class_weights[stage1_train_labels]
)

sampler = WeightedRandomSampler(
    weights=sample_weights,
    num_samples=len(sample_weights),
    replacement=True
)

print("Weighted sampler created successfully!")
```

```python
BATCH_SIZE = 16

stage1_train_loader = DataLoader(
    stage1_train,
    batch_size=BATCH_SIZE,
    sampler=sampler,
    num_workers=0
)

stage1_valid_loader = DataLoader(
    stage1_valid,
    batch_size=BATCH_SIZE,
    shuffle=False,
    num_workers=0
)

stage1_test_loader = DataLoader(
    stage1_test,
    batch_size=BATCH_SIZE,
    shuffle=False,
    num_workers=0
)

print("DataLoaders created!")
```

```python
weights = ResNet18_Weights.DEFAULT

model = resnet18(weights=weights)

print(model.fc)
```

```python
model.fc = nn.Linear(
    model.fc.in_features,
    2
)

model = model.to(device)

print(model.fc)
print("Model device:", device)
```

```python
criterion = nn.CrossEntropyLoss()

optimizer = torch.optim.AdamW(
    model.parameters(),
    lr=1e-4,
    weight_decay=1e-4
)

scheduler = torch.optim.lr_scheduler.ReduceLROnPlateau(
    optimizer,
    mode="max",
    factor=0.5,
    patience=2
)

print("Loss function, optimizer and scheduler created.")
```

```python
def train_one_epoch(model, loader, criterion, optimizer, device):
    
    model.train()
    
    running_loss = 0.0
    correct = 0
    total = 0
    
    for images, labels in loader:
        
        images = images.to(device)
        labels = labels.to(device)
        
        # Clear previous gradients
        optimizer.zero_grad()
        
        # Forward pass
        outputs = model(images)
        
        # Calculate loss
        loss = criterion(outputs, labels)
        
        # Backpropagation
        loss.backward()
        
        # Update weights
        optimizer.step()
        
        # Statistics
        running_loss += loss.item() * images.size(0)
        
        predictions = torch.argmax(outputs, dim=1)
        
        correct += (predictions == labels).sum().item()
        total += labels.size(0)
    
    epoch_loss = running_loss / total
    epoch_accuracy = correct / total
    
    return epoch_loss, epoch_accuracy
```

```python
def evaluate_model(model, loader, criterion, device):
    
    model.eval()
    
    running_loss = 0.0
    correct = 0
    total = 0
    
    all_labels = []
    all_predictions = []
    all_probabilities = []
    
    with torch.no_grad():
        
        for images, labels in loader:
            
            images = images.to(device)
            labels = labels.to(device)
            
            outputs = model(images)
            
            loss = criterion(outputs, labels)
            
            probabilities = torch.softmax(outputs, dim=1)
            
            predictions = torch.argmax(outputs, dim=1)
            
            running_loss += loss.item() * images.size(0)
            
            correct += (predictions == labels).sum().item()
            total += labels.size(0)
            
            all_labels.extend(labels.cpu().numpy())
            all_predictions.extend(predictions.cpu().numpy())
            
            # Probability of Diseased class
            all_probabilities.extend(
                probabilities[:, 1].cpu().numpy()
            )
    
    epoch_loss = running_loss / total
    epoch_accuracy = correct / total
    
    return (
        epoch_loss,
        epoch_accuracy,
        np.array(all_labels),
        np.array(all_predictions),
        np.array(all_probabilities)
    )
```

```python
NUM_EPOCHS = 10

best_val_accuracy = 0.0
best_model_state = None

history = {
    "train_loss": [],
    "train_accuracy": [],
    "val_loss": [],
    "val_accuracy": []
}

for epoch in range(NUM_EPOCHS):
    
    print(f"\nEpoch {epoch + 1}/{NUM_EPOCHS}")
    print("-" * 50)
    
    # Training
    train_loss, train_accuracy = train_one_epoch(
        model,
        stage1_train_loader,
        criterion,
        optimizer,
        device
    )
    
    # Validation
    val_loss, val_accuracy, _, _, _ = evaluate_model(
        model,
        stage1_valid_loader,
        criterion,
        device
    )
    
    # Scheduler
    scheduler.step(val_accuracy)
    
    # Save history
    history["train_loss"].append(train_loss)
    history["train_accuracy"].append(train_accuracy)
    history["val_loss"].append(val_loss)
    history["val_accuracy"].append(val_accuracy)
    
    # Save best model
    if val_accuracy > best_val_accuracy:
        
        best_val_accuracy = val_accuracy
        
        best_model_state = copy.deepcopy(
            model.state_dict()
        )
        
        print("★ New best validation model saved!")
    
    print(f"Train Loss     : {train_loss:.4f}")
    print(f"Train Accuracy : {train_accuracy:.4f}")
    print(f"Val Loss       : {val_loss:.4f}")
    print(f"Val Accuracy   : {val_accuracy:.4f}")
    print(f"Learning Rate  : {optimizer.param_groups[0]['lr']:.7f}")
```

```python
model.load_state_dict(best_model_state)

print(
    f"Best validation accuracy: "
    f"{best_val_accuracy:.4f}"
)
```

```python
stage2_model = resnet18(weights=None)

stage2_model.fc = nn.Linear(
    stage2_model.fc.in_features,
    2
)

checkpoint = torch.load(
    "stage1_pretrained.pt",
    map_location=device
)

stage2_model.load_state_dict(
    checkpoint["model_state_dict"]
)

stage2_model = stage2_model.to(device)

print("Stage 1 weights loaded into Stage 2 model.")
```

```python
for param in stage2_model.parameters():
    param.requires_grad = False

# Unfreeze the final convolutional block
for param in stage2_model.layer4.parameters():
    param.requires_grad = True

# Unfreeze classifier
for param in stage2_model.fc.parameters():
    param.requires_grad = True

print("Trainable layers:")
for name, param in stage2_model.named_parameters():
    if param.requires_grad:
        print(name)
```

```python
stage2_criterion = nn.CrossEntropyLoss()

stage2_optimizer = torch.optim.AdamW(
    filter(
        lambda p: p.requires_grad,
        stage2_model.parameters()
    ),
    lr=1e-5,
    weight_decay=1e-4
)

stage2_scheduler = torch.optim.lr_scheduler.ReduceLROnPlateau(
    stage2_optimizer,
    mode="max",
    factor=0.5,
    patience=2
)

print("Stage 2 optimizer ready.")
```

```python
STAGE2_EPOCHS = 10

best_stage2_val_accuracy = 0.0
best_stage2_state = None

stage2_history = {
    "train_loss": [],
    "train_accuracy": [],
    "val_loss": [],
    "val_accuracy": []
}

for epoch in range(STAGE2_EPOCHS):
    
    print(f"\nStage 2 Epoch {epoch + 1}/{STAGE2_EPOCHS}")
    print("-" * 55)
    
    # -------------------------
    # Training
    # -------------------------
    train_loss, train_accuracy = train_one_epoch(
        stage2_model,
        stage2_train_loader,
        stage2_criterion,
        stage2_optimizer,
        device
    )
    
    # -------------------------
    # Validation
    # -------------------------
    val_loss, val_accuracy, _, _, _ = evaluate_model(
        stage2_model,
        stage2_valid_loader,
        stage2_criterion,
        device
    )
    
    # Learning-rate scheduler
    stage2_scheduler.step(val_accuracy)
    
    # Save history
    stage2_history["train_loss"].append(train_loss)
    stage2_history["train_accuracy"].append(train_accuracy)
    stage2_history["val_loss"].append(val_loss)
    stage2_history["val_accuracy"].append(val_accuracy)
    
    # Save best model
    if val_accuracy > best_stage2_val_accuracy:
        
        best_stage2_val_accuracy = val_accuracy
        
        best_stage2_state = copy.deepcopy(
            stage2_model.state_dict()
        )
        
        print("★ New best Stage 2 model saved!")
    
    print(f"Train Loss     : {train_loss:.4f}")
    print(f"Train Accuracy : {train_accuracy:.4f}")
    print(f"Val Loss       : {val_loss:.4f}")
    print(f"Val Accuracy   : {val_accuracy:.4f}")
    print(
        f"Learning Rate  : "
        f"{stage2_optimizer.param_groups[0]['lr']:.7f}"
    )
```
