## 1. INTRODUCTION
Online toxicity poses severe risks to user wellbeing and community engagement.
This notebook analyzes the benchmark multilabel toxic comment classification dataset containing **159,571 comments** labeled across 6 distinct toxicity categories:
- 	oxic
- severe_toxic
- obscene
- 	hreat
- insult
- identity_hate

```python
import os
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
import warnings
warnings.filterwarnings('ignore')
sns.set_theme(style='whitegrid')
%matplotlib inline
```

```python
dataset_file = os.path.join('..', 'datasets', 'Original_Dataset(Toxic_Comment_Classification).csv')
if not os.path.exists(dataset_file):
    dataset_file = os.path.join('..', 'src', 'dataset', 'train.csv')

df_bench = pd.read_csv(dataset_file, nrows=30000)
print('Dataset Shape (sampled for rapid analysis):', df_bench.shape)
df_bench.head(3)
```

```python
labels = ['toxic', 'severe_toxic', 'obscene', 'threat', 'insult', 'identity_hate']
label_counts = df_bench[labels].sum()

plt.figure(figsize=(9, 5))
sns.barplot(x=label_counts.index, y=label_counts.values, palette='mako')
plt.title('Frequency of Toxicity Labels in Benchmark Dataset', fontsize=14, fontweight='bold')
plt.ylabel('Number of Occurrences')
plt.xticks(rotation=25)
plt.show()
```

```python
df_bench['char_length'] = df_bench['comment_text'].astype(str).apply(len)
df_bench['word_count'] = df_bench['comment_text'].astype(str).apply(lambda x: len(x.split()))

plt.figure(figsize=(12, 4))
plt.subplot(1, 2, 1)
sns.histplot(df_bench[df_bench['char_length'] < 1000]['char_length'], bins=40, color='#3498db', kde=True)
plt.title('Character Length Distribution (< 1000 chars)')

plt.subplot(1, 2, 2)
sns.boxplot(data=df_bench, x='toxic', y='word_count', palette={0: '#2ecc71', 1: '#e74c3c'})
plt.title('Word Count: Non-Toxic (0) vs Toxic (1)')
plt.ylim(0, 200)
plt.tight_layout()
plt.show()
```

## 2. MULTILABEL CORRELATION & CO-OCCURRENCE

```python
plt.figure(figsize=(8, 6))
sns.heatmap(df_bench[labels].corr(), annot=True, cmap='YlOrRd', fmt='.2f')
plt.title('Label Co-Occurrence Correlation Matrix', fontsize=14, fontweight='bold')
plt.show()
```

```python
clean_count = (df_bench[labels].sum(axis=1) == 0).sum()
toxic_any_count = (df_bench[labels].sum(axis=1) > 0).sum()

plt.figure(figsize=(6, 5))
plt.pie([clean_count, toxic_any_count], labels=['Clean Comments', 'Flagged as Toxic'], autopct='%1.1f%%', colors=['#27ae60', '#c0392b'], explode=(0, 0.1))
plt.title('Dataset Class Balance Overview', fontsize=14, fontweight='bold')
plt.show()
```

## 3. SUMMARY
- High class imbalance: over 89% of comments are clean.
- Severe toxicity, threats, and identity hate represent critical minority classes requiring targeted attention.
- Strong correlation exists between 	oxic, obscene, and insult categories.
